```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


temperatura = np.array([30, 25, 36, 18, 25, 29, 30, 33, 37, 31, 26, 37, 29, 26, 30, 31, 34, 38])

numero_sorvetes = np.array([20, 12, 50, 10, 18, 25, 26, 32, 48, 22, 16, 52, 24, 20, 28, 29, 35, 40])

df = pd.DataFrame({'temperatura': temperatura, 'numero_sorvetes': numero_sorvetes})

df.head()
#df.describe()
```

## Gerando o grafico

```python
plt.plot(df['temperatura'], df['numero_sorvetes'], '*')
plt.xlabel('Temperatura')
plt.ylabel('Sorvetes')
plt.show()
```

##Variavel dependente e independente

```python
from sklearn.model_selection import train_test_split

x = df['temperatura'].to_numpy()
y = df['numero_sorvetes'].to_numpy()

x_treino, x_teste, y_treino, y_teste = train_test_split(x, y, test_size=0.2)
```

Treinando modelo

```python
from sklearn.linear_model import LinearRegression

# define um modelo de regressão
modelo = LinearRegression()

# treina o modelo com os dados de treino
modelo.fit(x_treino.reshape(-1, 1), y_treino.reshape(-1, 1))

# y_previsto contém a previsão dos preços do nosso modelo
y_previsto = modelo.predict(x_teste.reshape(-1,1))

plt.plot(range(y_previsto.shape[0]), y_previsto, 'r--')
plt.plot(range(y_teste.shape[0]), y_teste, 'g--')
plt.legend(['Sorvetes previstos', 'Sorvetes vendidos'])
plt.xlabel('índice')
plt.ylabel('Sorvetes')
plt.show()
```

1. Viés Alto e UnderfittingO Gráfico: Mostra pontos de dados dispersos em formato de curva ("U") e um modelo representado por uma reta vermelha. A distância (d) entre os dados reais e a reta é muito grande.O Conceito: Uma reta representa uma equação simples (y = ax + b). Quando o modelo é simples demais, ele possui Viés Alto e não consegue aprender o padrão dos dados, gerando o fenômeno do Underfitting (erros tanto no treino quanto nos testes).

2. Viés Baixo (Modelo Ideal no Treino)O Gráfico: Os pontos azuis de dados estão perfeitamente distribuídos ao longo de uma curva parabólica vermelha em formato de "U" (y = x²).O Conceito: Ao usar um modelo mais complexo e adequado, a distância entre a linha e os pontos diminui drasticamente, caracterizando um Viés Baixo. O modelo consegue representar bem o conjunto de dados.

3. Variância Alta e OverfittingOs Gráficos: Dividem-se em duas etapas:Etapa de Treino: A curva ondulada vermelha passa exatamente por cima de todos os pontos azuis, "decorando-os".Etapa de Testes: Quando novos dados (pontos verdes) são apresentados, a curva vermelha erra a maioria deles por estar muito distante.O Conceito: Se o viés for excessivamente baixo, o modelo se torna complexo demais e memoriza a base de treino. Isso gera Variância Alta, fazendo com que ele perca a capacidade de generalizar para novos dados, resultando em Overfitting.

4. Validação Cruzada (Cross-Validation)O Esquema: Ilustra um conjunto de dados dividido em 4 blocos horizontais ao longo de 4 rodadas. Em cada rodada, um bloco diferente (cor laranja) é separado para Dados de Teste, enquanto os demais (cor branca) servem como Dados de Treino.O Conceito: Para solucionar o dilema do Tradeoff Viés/Variância, o documento cita técnicas como a redução de dimensionalidade e a validação cruzada. O teste repetido com partes diferentes dos dados garante que o modelo seja validado de forma robusta e equilibrada.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split

# 1. Gerar dados sintéticos baseados em uma parábola (y = x² + ruído)
np.random.seed(42)
X = np.sort(np.random.rand(40) * 10)[:, np.newaxis]
y = 0.5 * (X ** 2).ravel() + np.random.randn(40) * 3  # Dados reais com ruído

# Dividir os dados em Treino e Teste (conceito de Validação Cruzada/Divisão)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Criar pontos na linha X para desenhar as curvas dos modelos no gráfico
X_plot = np.linspace(0, 10, 100)[:, np.newaxis]

# Graus dos polinômios que representam a complexidade do modelo
graus = [1, 2, 15]
titulos = ['Grau 1: Underfitting (Viés Alto)', 'Grau 2: Modelo Ideal (Equilibrado)', 'Grau 15: Overfitting (Variância Alta)']

plt.figure(figsize=(18, 5))

for i, grau in enumerate(graus):
    ax = plt.subplot(1, 3, i + 1)

    # Criar e treinar o modelo polinomial
    modelo = make_pipeline(PolynomialFeatures(grau), LinearRegression())
    modelo.fit(X_train, y_train)

    # Fazer previsões na linha contínua para o gráfico
    y_plot = modelo.predict(X_plot)

    # Plotar os pontos de Treino e Teste
    plt.scatter(X_train, y_train, color='blue', label='Dados de Treino', alpha=0.7)
    plt.scatter(X_test, y_test, color='green', label='Dados de Teste', alpha=0.7)

    # Plotar a curva que o modelo aprendeu
    plt.plot(X_plot, y_plot, color='red', linewidth=2, label='Modelo (Previsão)')

    plt.title(titulos[i])
    plt.xlabel('X')
    plt.ylabel('y')
    plt.ylim(-10, 60)
    plt.legend()

plt.tight_layout()
plt.show()
```
